###**Outliers**

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Berdasarkan hasil pengecekan outlier pada seluruh kolom numerik, tidak ditemukan outlier pada kolom Price_USD maupun Units_Sold. Namun, pada kolom Revenue_USD ditemukan persentase outlier sebesar 0,4%. Jumlah tersebut tergolong sangat kecil sehingga tidak terlalu memengaruhi distribusi data secara keseluruhan. Outlier yang muncul kemungkinan berasal dari transaksi dengan nilai revenue yang jauh lebih tinggi dibandingkan transaksi lainnya. Selain itu, outlier tersebut masih dapat dianggap wajar karena kemungkinan sesuai dengan kondisi transaksi di lapangan, seperti adanya pembelian dalam jumlah besar atau produk dengan harga yang lebih tinggi. #**Exploratory Data Analysis (EDA)**

##**Comparison/Perbandingan**

**Aktivitas:** Menganalisis total revenue berdasarkan brand sepatu pada dataset penjualan sepatu.

**Tujuan:** Mengetahui brand sepatu yang memberikan kontribusi revenue terbesar terhadap perusahaan sehingga dapat membantu dalam memahami preferensi pelanggan serta mendukung pengambilan keputusan terkait strategi pemasaran dan pengelolaan produk secara lebih optimal.

**Visualisasi:** Bar chart.


```python
import matplotlib.pyplot as plt
import seaborn as sns

brand_sales = df.groupby('Brand')['Revenue_USD'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot( x=brand_sales.index, y=brand_sales.values, hue=brand_sales.index, palette='magma', legend=False )

plt.title('Total Revenue Berdasarkan Brand')
plt.xlabel('Brand')
plt.ylabel('Total Revenue (USD)')
plt.xticks(rotation=45)

plt.show()
```

Berdasarkan visualisasi total revenue berdasarkan brand, diketahui bahwa Skechers menjadi brand dengan kontribusi revenue tertinggi dibandingkan brand lainnya. Hal ini menunjukkan bahwa produk dari Skechers memiliki tingkat penjualan yang paling baik serta cukup diminati oleh pelanggan.

Selain itu, New Balance dan Adidas juga menunjukkan performa revenue yang relatif tinggi dengan selisih yang tidak terlalu jauh dari Skechers. Sementara itu, Nike menjadi brand dengan total revenue paling rendah pada dataset ini dibandingkan brand lainnya.

Perbedaan revenue antar brand menunjukkan adanya variasi preferensi pelanggan terhadap masing-masing merek sepatu. Oleh karena itu, perusahaan dapat mempertimbangkan untuk mempertahankan strategi pemasaran pada brand dengan performa tinggi serta mengevaluasi strategi penjualan pada brand dengan revenue yang lebih rendah agar penjualannya dapat lebih optimal.


##**Composition/Komposisi**

**Aktivitas:** Menganalisis proporsi revenue berdasarkan sales channel pada dataset penjualan sepatu.

**Tujuan:** Mengetahui sales channel dengan kontribusi revenue terbesar sehingga perusahaan dapat memahami channel penjualan yang paling efektif dan menentukan strategi pemasaran maupun distribusi produk secara lebih optimal.

**Visualisasi:** Pie chart.


```python
import matplotlib.pyplot as plt

variabel = df.groupby('Sales_Channel')['Revenue_USD'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,8))

plt.pie(
    variabel.values,
    labels=variabel.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Set3.colors
)

plt.title('Proporsi Revenue Berdasarkan Sales Channel')

plt.axis('equal')

plt.show()
```

Berdasarkan visualisasi proporsi revenue berdasarkan sales channel, diketahui bahwa Retail Store menjadi saluran distribusi dengan kontribusi revenue tertinggi dibandingkan saluran lainnya, yaitu sebesar 34.8%. Hal ini menunjukkan bahwa toko fisik mandiri masih menjadi pilihan utama pelanggan dalam melakukan transaksi dan memiliki efektivitas penjualan yang paling baik.

Selain itu, Mall dan Online juga menunjukkan performa revenue yang sangat kompetitif dengan kontribusi masing-masing sebesar 33.0% dan 32.2%. Selisih persentase yang tipis di antara ketiga saluran ini mengindikasikan bahwa distribusi pendapatan perusahaan sudah cukup merata dan tidak terpaku pada satu metode penjualan saja.

##**Distribution/Distribusi**

**Aktivitas:** Melihat distribusi nilai revenue penjualan pada dataset penjualan sepatu.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.histplot( df['Revenue_USD'], bins=20, kde=True, color='skyblue')

plt.title('Distribusi Revenue Penjualan')
plt.xlabel('Revenue (USD)')
plt.ylabel('Frequency')

plt.show()
```

Grafik ini juga menunjukkan pola distribusi yang melandai ke arah kanan (right-skewed distribution). Oleh karena itu, perusahaan dapat mempertimbangkan strategi untuk meningkatkan average order value (AOV) melalui teknik upselling atau bundling, sekaligus tetap menjaga ketersediaan stok pada kategori produk yang paling sering dibeli oleh pelanggan di rentang revenue rendah agar stabilitas pendapatan tetap terjaga.

##**Relationship/Hubungan**

**Aktivitas:** Menganalisis hubungan antara variabel Harga Produk, Jumlah Terjual, dan Revenue berdasarkan nilai korelasi pada heatmap.

**Tujuan:** Mengetahui seberapa kuat hubungan antara harga produk, jumlah unit terjual, dan revenue penjualan sehingga dapat membantu memahami faktor yang paling berpengaruh terhadap peningkatan revenue perusahaan.

**Visualisasi:** Correlation heatmap.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

df_indo = df.rename(columns={
    'Price_USD': 'Harga Produk',
    'Units_Sold': 'Jumlah Terjual',
    'Revenue_USD': 'Revenue'
})

sns.heatmap(
    df_indo[['Harga Produk', 'Jumlah Terjual', 'Revenue']].corr(),
    annot=True,
    cmap='viridis',
    fmt='.2f'
)

plt.title('Correlation Heatmap Variabel Numerik')

plt.show()
```

Berdasarkan grafik heatmap tersebut, diketahui bahwa variabel Jumlah Terjual memiliki korelasi positif yang paling kuat terhadap Revenue dengan nilai koefisien sebesar 0.73. Hal ini menunjukkan bahwa volume penjualan produk memiliki pengaruh yang lebih signifikan dalam meningkatkan total pendapatan dibandingkan faktor harga satuan produk saja.

Selain itu, variabel Harga Produk juga menunjukkan korelasi positif yang cukup berarti terhadap Revenue dengan nilai 0.59. Meskipun kontribusinya tidak sebesar jumlah unit yang terjual, peningkatan harga produk tetap berbanding lurus dengan peningkatan pendapatan, yang menandakan bahwa produk pada kategori harga yang lebih tinggi masih mampu memberikan kontribusi profitabilitas yang stabil.

Menariknya, korelasi antara Harga Produk dan Jumlah Terjual hampir tidak ada (mendekati nol, yaitu -0.02). Hal ini mengindikasikan bahwa tingkat harga tidak secara langsung menghambat minat beli pelanggan dalam dataset ini, sehingga fluktuasi harga tidak menyebabkan penurunan drastis pada jumlah unit yang laku. Oleh karena itu, perusahaan dapat mempertimbangkan strategi untuk fokus pada peningkatan volume penjualan melalui berbagai kanal, sambil tetap menjaga keseimbangan harga produk untuk mengoptimalkan total revenue secara keseluruhan.
